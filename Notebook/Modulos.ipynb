```python
#Un modulo contiene funciones y atributos que pueden ser reutilizados en otro codigo
#Tambien puede contener otros modulos, esto en proyectos mas grandes
#Python tiene mas de 200 modulos integrados, entre esos OS y STRING
```

```python
#OS MODULE: Usado para interactuar con el sistema operativo de la computadora

#Importamos el modulo
import os
print(type(os))

#Para saber los distintos comandos del modulo usamos la funcion help()
print(help(os))
```

```python
#OS MODULE
#Para usar una funcion del modulo ponemos os.namefunc()
print(os.getcwd())
#Podemos guardarlo en variables
work_directory = os.getcwd()
print(f'Directorio actual: {work_directory}')

#Para movernos del directorio usamos os.chdir(direccion/del/directorio)
new_directory = os.chdir("c:\\Users\\dylan\\Documents\\Code")
print(f'Nuevo directorio: {os.getcwd()}')

#RECUERDA USAR print(help(os)) PARA VER LOS OTROS COMDANDOS DEL MODULO
```

```python
#string Module: Simplifica el procesamiento de texto
import string

#algunas funciones
print(string.ascii_lowercase)
print(string.digits)
print(string.punctuation)

#EJEMPLO

word = 'chocoLATE'

print(''.join(i for i in word if i in string.ascii_lowercase))
print(''.join(i for i in word if i in string.ascii_uppercase))
```
